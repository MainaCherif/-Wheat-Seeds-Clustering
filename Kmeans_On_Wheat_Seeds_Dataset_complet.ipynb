### <font color = 'gold'> DBSCAN

```python
# Importer le modèle DBSCAN
from sklearn.cluster import DBSCAN
```

##### <font color = 'bisque'> Recherche des paramètres optimaux

```python
# Recherche des paramètres optimaux (eps, min_samples) : on maximise la silhouette
# tout en gardant un nombre d'anomalies raisonnable (< 25% des points)
from sklearn.metrics import silhouette_score as _sil
best = None
for eps in np.linspace(0.5, 2.0, 16):
  for min_sam in range(3, 12):
    db_ = DBSCAN(eps=eps, min_samples=min_sam).fit(x)
    labels_ = db_.labels_
    n_noise = (labels_ == -1).sum()
    n_clusters_ = len(set(labels_) - {-1})
    if n_clusters_ < 2 or n_clusters_ > 6 or n_noise > 0.25 * len(x):
      continue
    score = _sil(x[labels_ != -1], labels_[labels_ != -1])
    if best is None or score > best[0]:
      best = (score, round(eps, 3), min_sam)
print("Meilleur (silhouette, eps, min_samples) trouvé :", best)
```

```python
# Instantier le modèle avec les paramètres optimaux trouvés (eps = 0.9, min_samples = 8)
dbscan_model = DBSCAN(eps=0.9, min_samples=8)
# Entrainer le modèle
dbscan_model.fit(x)
```

```python
# Prédire la partition
dbscan_partition = dbscan_model.labels_
dbscan_partition[:20]
```

```python
# Nombre d'observations dans chaque classe
val_dbscan = pd.Series(dbscan_partition).value_counts()
val_dbscan
```

```python
# Diagramme en barre des nombres d'observations
bar_plot(val_dbscan)
```

```python
# Visualiser les classes
scatter_plot(df_pca, dbscan_partition)
```

```python
# les anomalies (graines qui ne ressemblent à aucune classe)
anomalies = df.index[dbscan_partition == -1]
anomalies
```

```python
# partition sans les anomalies (-1)
mask_sans_anomalies = dbscan_partition != -1
x_sans_anomalies = x[mask_sans_anomalies]
dbscan_partition_sans_anomalies = dbscan_partition[mask_sans_anomalies]
df_pca_sans_anomalies = df_pca[mask_sans_anomalies].reset_index(drop=True)
```

```python
# visualiser les classes sans les anomalies
scatter_plot(df_pca_sans_anomalies, dbscan_partition_sans_anomalies)
```

##### <font color = 'bisque'> Indice de silhouette de la partition du DBSCAN

### <font color = 'gold'> Déploiement du meilleur modèle (DBSCAN)

La comparaison des indices de silhouette montre que **DBSCAN** fournit la meilleure partition.
C'est donc ce modèle que nous allons déployer.

**Objectif du déploiement** : saisir les 7 caractéristiques morphologiques d'une **nouvelle graine**
et obtenir automatiquement la **classe** à laquelle elle appartient (parmi les classes de la partition).

#### <font color = 'teal'> 1. Comprendre le problème : DBSCAN n'a pas de méthode `predict()`

Avec KMeans, on peut écrire `kmeans_model.predict(nouvelle_graine)` : chaque classe possède un centre,
il suffit de trouver le centre le plus proche.

DBSCAN, lui, **ne calcule pas de centres**. Il n'a donc pas de méthode `predict()` : il sait seulement
classer les points qu'il a vus pendant l'entrainement (`dbscan_model.labels_`).

**La solution, très simple** : DBSCAN mémorise les **points cœur** (les points denses autour desquels
les classes se sont formées). Ils sont stockés dans `dbscan_model.components_`.
Pour une nouvelle graine, on applique exactement la règle de DBSCAN :

1. on cherche le **point cœur le plus proche** de la nouvelle graine ;
2. si la distance est **inférieure ou égale à `eps`** (ici 0.9), la graine reçoit la **classe de ce point cœur** ;
3. sinon, la graine est trop loin de toutes les zones denses : c'est une **anomalie** (classe `-1`).

#### <font color = 'teal'> 2. Récupérer les points cœur de DBSCAN

```python
# Les points cœur mémorisés par DBSCAN
points_coeur = dbscan_model.components_
# La classe de chacun de ces points cœur
labels_coeur = dbscan_model.labels_[dbscan_model.core_sample_indices_]

print("Nombre de points cœur :", len(points_coeur))
print("Classes présentes      :", np.unique(labels_coeur))
```

#### <font color = 'teal'> 3. La fonction de prédiction

Contrairement au notebook Customer_Segmentation, ici les données **n'ont pas été normalisées**
(`x = df.values`, sans `normalize()`). Il ne faut donc **pas** normaliser la nouvelle graine non plus :
on applique exactement le même prétraitement (aucun) que sur les données d'entrainement.

```python
def predire_classe(nouvelle_graine):

    # Mettre la graine sous forme de matrice à 1 ligne
    graine = np.array(nouvelle_graine, dtype=float).reshape(1, -1)

    # Distances entre la graine et tous les points cœur
    distances = np.linalg.norm(points_coeur - graine, axis=1)

    # Le point cœur le plus proche
    plus_proche = distances.argmin()

    # Règle de DBSCAN : dans le voisinage (eps) ou pas ?
    if distances[plus_proche] <= dbscan_model.eps:
        # classe du point cœur le plus proche
        return labels_coeur[plus_proche]
    else:
        # anomalie
        return -1
```

```python
# Test rapide
print("Classe prédite  :", predire_classe(df.iloc[0].values))
print("Classe de DBSCAN:", dbscan_partition[0])
```
